6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math

class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()

class Cylinder(Figure3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * (self.radius ** 2) * self.height

class Cube(Figure3D):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side ** 3

fig1 = Cylinder(radius=5, height=10)
fig2 = Cube(side=4)

print(f"Cylinder: {fig1.get_volume():.2f}")
print(f"Cube: {fig2.get_volume():.2f}")

result = fig1 + fig2

print(f"Result: {result:.2f}")
```

7. (3б). Продемонструйте різницю між isinstance та issubclass.

```python
import math

class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()

class Cylinder(Figure3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * (self.radius ** 2) * self.height

class Cube(Figure3D):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side ** 3

cube = Cube(side=4)

print(isinstance(cube, Cube)) # Об'єкт належить класу
print(isinstance(cube, Figure3D)) # Об'єкт належить нащадку
print(issubclass(Cube, Figure3D)) # Cube успадкований від Figure3D
print(issubclass(Figure3D, Cube)) # Figure3D не є нащадком Cube
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, wallet_id, initial_balance=0):
        self.__id = wallet_id
        self.__balance = initial_balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            print("Balance cannot be negative")
        else:
            self.__balance = value

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Wrong amount to reduce")

wallet = EWallet(1, 100)
print(f"Balance: {wallet.balance}")
wallet.deposit(50)
print(f"Balance: {wallet.balance}")
wallet.balance = 200
print(f"Balance: {wallet.balance}")
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class ShoppingCart:
    def __init__(self):
        self.items = []

    def addItem(self, item):
        self.items.append(item)

    def removeItem(self, item_name):
        for item in self.items:
            if item.name == item_name:
                self.items.remove(item)
                return
        print("Item was not found")

    def showItems(self):
        if not self.items:
            print("Cart is empty")
        else:
            for item in self.items:
                print(f"{item.name}: {item.price} UAH")

cart = ShoppingCart()
item1 = Item("Sugar", 20)
item2 = Item("Flour", 35)

cart.addItem(item1)
cart.addItem(item2)
cart.showItems()
cart.removeItem("Sugar")
cart.showItems()
```

Створіть клас Car, який успадковує обидва класи Vehicle та Engine. Продемонструйте використання множинної спадковості, створивши обʼєкт класу Car та отримавши mro для цього класу.

```python
class Vehicle:
    def __init__(self, brand, production_age):
        self.brand = brand
        self.production_age = production_age

class Engine:
    def __init__(self, engine_type, size):
        self.engine_type = engine_type
        self.size = size

class Car(Vehicle, Engine):
    def __init__(self, brand, production_age, engine_type, size, model):
        Vehicle.__init__(self, brand, production_age)
        Engine.__init__(self, engine_type, size)
        self.model = model

    def displayInfo(self):
        print(f"Car: {self.brand} {self.model}")
        print(f"Engine: {self.size}L ({self.engine_type}), Age: {self.production_age}")

my_car = Car("BMW", 3, "petrol", 3.0, "M3")
my_car.displayInfo()

for cls in Car.mro():
    print(cls)
```

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekDayIterator:
    def __init__(self):
        self.days = [
            "Monday", "Tuesday", "Wednesday",
            "Tuesday", "Friday", "Saturday", "Sunday"
        ]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.days):
            day = self.days[self.index]
            self.index += 1
            return day
        else:
            raise StopIteration

week = WeekDayIterator()

for day in week:
    print(day)
```
